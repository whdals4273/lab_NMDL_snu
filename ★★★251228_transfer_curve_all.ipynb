## file_path, W_um/L_um, Cox_total_F update

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress

# Device parameters: change these values and all Vth / SS / mobility results update together
W_um = 40
L_um = 20
W_over_L = W_um / L_um
L_m = L_um * 1e-6
W_m = W_um * 1e-6

# Gate capacitance for this device. Update if a different device is measured.
Cox_total_F = 4.25e-12

file_path = '260206_I_V Sweep - Unit (transfer) [260206_ALD_IGZO_#6_WL_1010_R02_lower_TR_2nd(2) ; 2_6_2026 1_54_35 PM].csv'
with open(file_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

for i, line in enumerate(lines):
    if any(col.strip().startswith('DataValue') for col in line.split(',')):
        header_row = i - 2
        break
df = pd.read_csv(file_path, sep=',', header=header_row)

df = df.iloc[:, 1:]
df.columns = df.columns.str.strip()

df.head(10)
```

## data 제대로 들어왔는지 확인

```python
df['ID'] = df['ID'].abs()
df['ID'] = np.clip(df['ID'], 1e-16, None)

df.value_counts()
df.describe()
row_counts = df.shape[0]

for i in df.columns:
    empty = df[i].isna().sum()
    print(f'{i} column에서 빈값은 {empty}개임.')
```

```python
# 1. Transfer Curve (log scale: ID only)
fig, ax = plt.subplots()
for VD, g in df.groupby('VD'):
    ax.semilogy(g['VG'], g['ID'], label=fr'$V_D$ = {VD:.2f} [V]')
ax.set_xlabel('VG [V]')
ax.set_ylabel('ID [A]')
ax.set_title('Transfer Curve_log', fontsize=20)
ax.legend(title='Vd[V]', bbox_to_anchor=(1, 1), loc='upper left', fontsize=8)
ax.grid(True)
plt.tight_layout()

# 2. Transfer Curve_log normalized (ID / (W/L))
fig, ax = plt.subplots()
for VD, g in df.groupby('VD'):
    ax.semilogy(g['VG'], g['ID'] / W_over_L, label=fr'$V_D$ = {VD:.2f} [V]')
ax.set_xlabel('VG [V]')
ax.set_ylabel('ID / (W/L)  [A]')
ax.set_title('Transfer Curve_log  (normalized by W/L)', fontsize=20)
ax.legend(title='Vd[V]', bbox_to_anchor=(1, 1), loc='upper left', fontsize=8)
ax.grid(True)
plt.tight_layout()

# 3. Transfer Curve_log — ID & |IG| on same y-axis
fig, ax = plt.subplots(figsize=(8, 5))
colors_list = plt.cm.tab10.colors
for idx, (VD, g) in enumerate(df.groupby('VD')):
    c = colors_list[idx % len(colors_list)]
    ax.semilogy(g['VG'], g['ID'], color=c, label=fr'ID, $V_D$={VD:.2f} V')
    ax.semilogy(g['VG'], g['IG'].abs().clip(lower=1e-16), linestyle='--', color=c, alpha=0.7,
                label=fr'|IG|, $V_D$={VD:.2f} V')
ax.set_xlabel('VG [V]')
ax.set_ylabel('Current [A]')
ax.set_title('Transfer Curve_log  (ID: solid, |IG|: dashed, same color = same VD)', fontsize=12)
ax.legend(bbox_to_anchor=(1, 1), loc='upper left', fontsize=7)
ax.grid(True)
plt.tight_layout()

# 4. Transfer Curve_log — ID & |IS| on same y-axis
fig, ax = plt.subplots(figsize=(8, 5))
for idx, (VD, g) in enumerate(df.groupby('VD')):
    c = colors_list[idx % len(colors_list)]
    ax.semilogy(g['VG'], g['ID'], color=c, label=fr'ID, $V_D$={VD:.2f} V')
    ax.semilogy(g['VG'], g['IS'].abs().clip(lower=1e-16), linestyle='--', color=c, alpha=0.7,
                label=fr'|IS|, $V_D$={VD:.2f} V')
ax.set_xlabel('VG [V]')
ax.set_ylabel('Current [A]')
ax.set_title('Transfer Curve_log  (ID: solid, |IS|: dashed, same color = same VD)', fontsize=12)
ax.legend(bbox_to_anchor=(1, 1), loc='upper left', fontsize=7)
ax.grid(True)
plt.tight_layout()

# 5. Transfer Curve (linear scale: ID)
fig, ax = plt.subplots()
for VD, g in df.groupby('VD'):
    ax.plot(g['VG'], g['ID'], label=fr'$V_D$ = {VD:.2f} [V]')
ax.set_xlabel('VG [V]')
ax.set_ylabel('ID [A]')
ax.set_title('Transfer Curve_linear', fontsize=20)
ax.legend(title='Vd[V]', bbox_to_anchor=(1, 1), loc='upper left', fontsize=8)
ax.grid(True)
plt.tight_layout()

# 6. VG vs |IG| (log scale)
fig, ax = plt.subplots()
for VD, g in df.groupby('VD'):
    ax.semilogy(g['VG'], g['IG'].abs().clip(lower=1e-16), label=fr'$V_D$ = {VD:.2f} [V]')
ax.set_xlabel('VG [V]')
ax.set_ylabel('|IG| [A]')
ax.set_title('VG vs |IG| (log scale)', fontsize=20)
ax.legend(title='Vd[V]', bbox_to_anchor=(1, 1), loc='upper left', fontsize=8)
ax.grid(True)
plt.tight_layout()

# 7. VG vs |IS+ID| & |IG| (log scale)
fig, ax = plt.subplots(figsize=(8, 5))
for idx, (VD, g) in enumerate(df.groupby('VD')):
    c = colors_list[idx % len(colors_list)]
    is_plus_id = (g['IS'] + g['ID']).abs().clip(lower=1e-16)
    ig_abs = g['IG'].abs().clip(lower=1e-16)
    ax.semilogy(g['VG'], is_plus_id, color=c, label=fr'|IS+ID|, $V_D$={VD:.2f} V')
    ax.semilogy(g['VG'], ig_abs, linestyle='--', color=c, alpha=0.7,
                label=fr'|IG|, $V_D$={VD:.2f} V')
ax.set_xlabel('VG [V]')
ax.set_ylabel('Current [A]')
ax.set_title('VG vs |IS+ID| & |IG|  (|IS+ID|: solid, |IG|: dashed, same color = same VD)', fontsize=11)
ax.legend(bbox_to_anchor=(1, 1), loc='upper left', fontsize=7)
ax.grid(True)
plt.tight_layout()

plt.show()
```

## get vth, ss

```python
target1 = -10  # log10(|ID|/(W/L)) = -10 -> Vth
target2 = -9   # log10(|ID|/(W/L)) = -9  -> SS upper decade

vth_by_vd_10 = {}
vth_by_vd_9 = {}
ss_by_vd = {}

for vd, g in df.groupby('VD'):
    g = g.sort_values('ID')

    VG = g['VG'].to_numpy(dtype=float)
    ID = g['ID'].to_numpy(dtype=float)

    absID = np.clip(np.abs(ID), 1e-300, None)
    normID = absID / W_over_L
    logI = np.log10(normID)

    diff1 = logI - target1
    diff2 = logI - target2
    cross_idx_10 = np.where(diff1[:-1] * diff1[1:] <= 0)[0]
    cross_idx_9 = np.where(diff2[:-1] * diff2[1:] <= 0)[0]

    if len(cross_idx_10) == 0:
        vth_by_vd_10[vd] = np.nan
        vth_by_vd_9[vd] = np.nan
        ss_by_vd[vd] = np.nan
        continue

    i1 = cross_idx_10[0]
    x0, x1 = VG[i1], VG[i1 + 1]
    y0, y1 = logI[i1], logI[i1 + 1]

    if y1 == y0:
        vth_10 = np.nan
    else:
        vth_10 = x0 + (target1 - y0) * (x1 - x0) / (y1 - y0)

    vth_by_vd_10[vd] = vth_10

    if len(cross_idx_9) == 0:
        vth_by_vd_9[vd] = np.nan
        ss_by_vd[vd] = np.nan
        continue

    i2 = cross_idx_9[0]
    z0, z1 = VG[i2], VG[i2 + 1]
    t0, t1 = logI[i2], logI[i2 + 1]

    if t1 == t0:
        vth_9 = np.nan
    else:
        vth_9 = z0 + (target2 - t0) * (z1 - z0) / (t1 - t0)

    vth_by_vd_9[vd] = vth_9

    if np.isnan(vth_10) or np.isnan(vth_9):
        ss_by_vd[vd] = np.nan
    else:
        ss_by_vd[vd] = ((vth_9 - vth_10) / (target2 - target1)) * 1000

print(f'W/L = {W_um:.1f} um / {L_um:.1f} um = {W_over_L:.2f}')
print(f'Vth criterion: log10(|ID|/(W/L)) = {target1}')
for vd in sorted(vth_by_vd_10.keys()):
    print(f'VD={vd}: Vth={vth_by_vd_10[vd]:.6f} V, SS={ss_by_vd[vd]:.3f} mV/dec')

summary_df = pd.DataFrame({
    'VD[V]': sorted(vth_by_vd_10.keys()),
    'Vth_normID=1e-10[V]': [vth_by_vd_10[vd] for vd in sorted(vth_by_vd_10.keys())],
    'SS[mV/dec]': [ss_by_vd[vd] for vd in sorted(ss_by_vd.keys())],
})
summary_df
```

## Field-Effect Mobility  (gm vs Vg 방법)

**원리**: linear 영역에서 $I_D = \frac{\mu_{FE} C_{ox} W}{L}(V_{GS}-V_{th}) \cdot V_{DS}$  
→ transconductance $g_m = \frac{dI_D}{dV_G}$로부터:

$$\mu_{FE} = \frac{g_m \cdot L^2}{C_{ox,total} \cdot V_{DS}}$$

$g_m$은 측정된 $I_D$–$V_G$ 곡선을 수치 미분하여 구하며, VG 각 점에서 $\mu_{FE}$를 계산하므로 **gate bias에 따른 mobility 변화**를 직접 관찰할 수 있음.

```python
# VG vs μ_FE (Field-Effect Mobility) plot
mobility_curve_rows = []

fig, ax = plt.subplots(figsize=(7, 5))
for vd, g in df.groupby('VD'):
    g = g.sort_values('VG').copy()
    vg = g['VG'].to_numpy(dtype=float)
    idrain = g['ID'].to_numpy(dtype=float)

    gm = np.gradient(idrain, vg)
    if vd == 0:
        mu_cm2_vs = np.full_like(gm, np.nan, dtype=float)
    else:
        mu_m2_vs = gm * (L_m ** 2) / (Cox_total_F * abs(vd))
        mu_cm2_vs = mu_m2_vs * 1e4

    plot_mu = np.where(mu_cm2_vs > 0, mu_cm2_vs, np.nan)
    ax.plot(vg, plot_mu, marker='o', markerfacecolor='none', label=f'Vd = {vd}[V]')

    temp_df = pd.DataFrame({
        'VG[V]': vg,
        'VD[V]': vd,
        'gm[A/V]': gm,
        'mu_FE[cm^2/Vs]': mu_cm2_vs,
    })
    mobility_curve_rows.append(temp_df)

ax.set_xlabel('VG[V]')
ax.set_ylabel(r'$\mu_{FE}$ [cm$^2$/Vs]')
ax.set_title(r'VG vs $\mu_{FE}$ (Field-Effect Mobility)', fontsize=20)
ax.legend(title='Vd[V]', bbox_to_anchor=(1, 1), loc='upper left', fontsize=8)
ax.grid(True)
plt.tight_layout()
plt.show()
```

```python
mobility_curve_df = pd.concat(mobility_curve_rows, ignore_index=True)
mobility_curve_df.head(20)
```

```python
mobility_summary_rows = []
for vd, g in mobility_curve_df.groupby('VD[V]'):
    positive_mu = g.loc[g['mu_FE[cm^2/Vs]'] > 0, 'mu_FE[cm^2/Vs]']
    mobility_summary_rows.append({
        'VD[V]': vd,
        'Vth_normID=1e-10[V]': vth_by_vd_10.get(vd, np.nan),
        'SS[mV/dec]': ss_by_vd.get(vd, np.nan),
        'mu_FE_max[cm^2/Vs]': positive_mu.max() if len(positive_mu) else np.nan,
        'mu_FE_mean[cm^2/Vs]': positive_mu.mean() if len(positive_mu) else np.nan,
    })

mobility_summary_df = pd.DataFrame(mobility_summary_rows).sort_values('VD[V]').reset_index(drop=True)
mobility_summary_df
```

## Saturation Region Mobility  (√|Id| vs Vg 방법)

**원리**: saturation 영역에서 $I_D = \frac{\mu_{sat} C_{ox} W}{2L}(V_{GS}-V_{th})^2$  
→ $\sqrt{|I_D|}$ vs $V_G$ 직선 fit의 기울기 $k$로부터:

$$\mu_{sat} = k^2 \cdot \frac{2L^2}{C_{ox,total}}$$

Fit 구간: $\frac{d\sqrt{|I_D|}}{dV_G}$ 가 최대값의 50 % 이상인 구간 (steepest turn-on region)

```python
sat_rows = []

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax_sqrt, ax_mu = axes
colors = plt.cm.tab10.colors

for idx, (vd, g) in enumerate(df.groupby('VD')):
    g      = g.sort_values('VG').copy()
    vg     = g['VG'].to_numpy(dtype=float)
    idr    = g['ID'].to_numpy(dtype=float)   # already abs-clipped

    sqrt_id = np.sqrt(idr)
    color   = colors[idx % len(colors)]
    ax_sqrt.plot(vg, sqrt_id, 'o', markerfacecolor='none',
                 label=f'Vd={vd:.2f}V', color=color)

    # fit region: d(√Id)/dVg ≥ 50 % of its peak value (steepest turn-on)
    d_sqrt  = np.gradient(sqrt_id, vg)
    peak_i  = int(np.argmax(d_sqrt))
    thresh  = 0.5 * d_sqrt[peak_i]
    mask    = (d_sqrt >= thresh) & (np.arange(len(vg)) <= peak_i + 2)

    if mask.sum() < 2 or d_sqrt[peak_i] <= 0:
        sat_rows.append({'VD[V]': vd, 'mu_sat[cm^2/Vs]': np.nan, 'Vth_sat[V]': np.nan, 'R2_sat': np.nan})
        continue

    slope, intercept, r_val, *_ = linregress(vg[mask], sqrt_id[mask])

    if slope <= 0:
        sat_rows.append({'VD[V]': vd, 'mu_sat[cm^2/Vs]': np.nan, 'Vth_sat[V]': np.nan, 'R2_sat': np.nan})
        continue

    # μ_sat [cm²/Vs]:  slope² × 2L² / Cox_total_F × 1e4
    mu_sat  = slope**2 * 2 * L_m**2 / Cox_total_F * 1e4
    vth_sat = -intercept / slope   # x-intercept of the fit line

    # draw extrapolated fit line (down to x-axis)
    vg_lo   = min(vg[mask].min(), vth_sat - 0.05)
    vg_hi   = vg[mask].max() + 0.1
    vg_line = np.linspace(vg_lo, vg_hi, 100)
    ax_sqrt.plot(vg_line, np.maximum(slope * vg_line + intercept, 0),
                 '--', color=color, lw=1.8)

    sat_rows.append({'VD[V]': vd, 'mu_sat[cm^2/Vs]': mu_sat,
                     'Vth_sat[V]': vth_sat, 'R2_sat': r_val**2})

ax_sqrt.set_xlabel('VG [V]')
ax_sqrt.set_ylabel(r'$\sqrt{|I_D|}$  [A$^{1/2}$]')
ax_sqrt.set_title(r'$\sqrt{|I_D|}$ vs $V_G$  (dashed = linear fit)', fontsize=13)
ax_sqrt.legend(title='Vd[V]', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=8)
ax_sqrt.grid(True)

# μ_sat bar chart vs VD
sat_df  = pd.DataFrame(sat_rows)
valid   = sat_df.dropna(subset=['mu_sat[cm^2/Vs]'])
ax_mu.bar(valid['VD[V]'].astype(str), valid['mu_sat[cm^2/Vs]'],
          color='steelblue', alpha=0.75, edgecolor='k', linewidth=0.6)
ax_mu.set_xlabel('VD [V]')
ax_mu.set_ylabel(r'$\mu_{sat}$  [cm$^2$/Vs]')
ax_mu.set_title(r'Saturation Mobility vs $V_D$', fontsize=13)
ax_mu.grid(True, axis='y')

plt.tight_layout()
plt.show()
```

## Linear Region Mobility  (Id vs Vg 직선 fit 방법)

**원리**: linear 영역에서 $I_D = \frac{\mu_{lin} C_{ox} W}{L}(V_{GS}-V_{th}) \cdot V_{DS}$  
→ $I_D$ vs $V_G$ 직선 fit의 기울기 $k$로부터:

$$\mu_{lin} = k \cdot \frac{L^2}{C_{ox,total} \cdot V_{DS}}$$

Fit 구간: $g_m = \frac{dI_D}{dV_G}$ 가 최대값의 50 % 이상인 구간 (steepest turn-on region)


```python
lin_rows = []

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax_id, ax_mu = axes
colors = plt.cm.tab10.colors

for idx, (vd, g) in enumerate(df.groupby('VD')):
    g = g.sort_values('VG').copy()
    vg = g['VG'].to_numpy(dtype=float)
    idr = g['ID'].to_numpy(dtype=float)

    color = colors[idx % len(colors)]
    ax_id.plot(vg, idr, 'o', markerfacecolor='none', label=f'Vd={vd:.2f}V', color=color)

    if abs(vd) < 1e-9:
        lin_rows.append({'VD[V]': vd, 'mu_lin[cm^2/Vs]': np.nan, 'Vth_lin_fit[V]': np.nan, 'R2_lin': np.nan})
        continue

    # fit region: gm >= 50% of gm_max (steepest turn-on in Id-Vg)
    gm = np.gradient(idr, vg)
    peak_i = int(np.argmax(gm))
    thresh = 0.5 * gm[peak_i]
    mask = (gm >= thresh) & (np.arange(len(vg)) <= peak_i + 2)

    if mask.sum() < 2 or gm[peak_i] <= 0:
        lin_rows.append({'VD[V]': vd, 'mu_lin[cm^2/Vs]': np.nan, 'Vth_lin_fit[V]': np.nan, 'R2_lin': np.nan})
        continue

    slope, intercept, r_val, *_ = linregress(vg[mask], idr[mask])

    if slope <= 0:
        lin_rows.append({'VD[V]': vd, 'mu_lin[cm^2/Vs]': np.nan, 'Vth_lin_fit[V]': np.nan, 'R2_lin': np.nan})
        continue

    # μ_lin [cm²/Vs]: slope = (W/L)·μ_lin·Cox·Vd  →  μ_lin = slope·L²/(Cox_total·Vd)·1e4
    mu_lin = slope * L_m**2 / (Cox_total_F * abs(vd)) * 1e4
    vth_lin_fit = -intercept / slope  # x-intercept of fit line

    vg_lo = min(vg[mask].min(), vth_lin_fit - 0.05)
    vg_hi = vg[mask].max() + 0.1
    vg_line = np.linspace(vg_lo, vg_hi, 100)
    ax_id.plot(vg_line, np.maximum(slope * vg_line + intercept, 0), '--', color=color, lw=1.8)

    lin_rows.append({'VD[V]': vd, 'mu_lin[cm^2/Vs]': mu_lin, 'Vth_lin_fit[V]': vth_lin_fit, 'R2_lin': r_val**2})

ax_id.set_xlabel('VG [V]')
ax_id.set_ylabel(r'$I_D$  [A]')
ax_id.set_title(r'$I_D$ vs $V_G$  (dashed = linear fit)', fontsize=13)
ax_id.legend(title='Vd[V]', bbox_to_anchor=(1.01, 1), loc='upper left', fontsize=8)
ax_id.grid(True)

lin_df = pd.DataFrame(lin_rows)
valid_lin = lin_df.dropna(subset=['mu_lin[cm^2/Vs]'])
ax_mu.bar(valid_lin['VD[V]'].astype(str), valid_lin['mu_lin[cm^2/Vs]'],
          color='coral', alpha=0.75, edgecolor='k', linewidth=0.6)
ax_mu.set_xlabel('VD [V]')
ax_mu.set_ylabel(r'$\mu_{lin}$  [cm$^2$/Vs]')
ax_mu.set_title(r'Linear Mobility ($\mu_{lin}$) vs $V_D$', fontsize=13)
ax_mu.grid(True, axis='y')

plt.tight_layout()
plt.show()
```

```python
# μ_FE + μ_lin + μ_sat 통합 요약 테이블
sat_summary_df = sat_df.merge(lin_df, on='VD[V]', how='left').merge(
    mobility_summary_df[['VD[V]', 'Vth_normID=1e-10[V]', 'SS[mV/dec]', 'mu_FE_max[cm^2/Vs]']],
    on='VD[V]', how='left'
).rename(columns={
    'Vth_normID=1e-10[V]': 'Vth_log[V]',
})[['VD[V]', 'Vth_sat[V]', 'Vth_lin_fit[V]', 'Vth_log[V]', 'SS[mV/dec]',
    'mu_sat[cm^2/Vs]', 'mu_lin[cm^2/Vs]', 'mu_FE_max[cm^2/Vs]',
    'R2_sat', 'R2_lin']]

print(f'W={W_um} um  L={L_um} um  Cox_total={Cox_total_F:.3e} F')
sat_summary_df
```
