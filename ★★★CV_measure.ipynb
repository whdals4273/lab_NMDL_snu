```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import linregress

# Device parameters: change these values and all Vth / SS / mobility results update together

file_path = 'igzo_leakage_data_260626/CV_meas/'+'[#2_jumbo_200C_R1]C-V with freq 1st [-3.00;0.05;3.00]~[0.03,CPD]-0.txt'
with open(file_path, 'r', encoding='utf-8') as f:
    lines = f.readlines()

df = pd.read_csv(file_path, sep='\t', header=0)


df.columns = df.columns.str.strip()

print(df.head(10))
```

```python
freq_cols = ['1.000k', '10.000k', '100.000k', '1.000M']
labels    = ['1 kHz', '10 kHz', '100 kHz', '1 MHz']
colors    = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

fig, ax = plt.subplots()

for col, label, color in zip(freq_cols, labels, colors):
    ax.plot(df['V'], df[col] * 1e12, label=label, color=color, lw=1.5)

ax.set_xlabel('Voltage (V)', fontsize=12)
ax.set_ylabel('Capacitance (pF)', fontsize=12)
ax.set_title('C-V Characteristics', fontsize=13)
ax.legend(bbox_to_anchor=(1, 1),title='Frequency', fontsize=10)
ax.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```
